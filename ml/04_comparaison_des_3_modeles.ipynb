# Comparaison des 3 modèles — Choix final MSPR

Ce notebook regroupe les résultats exportés par :
- `01_modele_linear_regression.ipynb`
- `02_modele_ridge.ipynb`
- `03_modele_random_forest.ipynb`

**Règle de choix :** le modèle avec la **RMSE la plus faible** est retenu (POC participation).

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..").resolve()
DM = ROOT / "outputs" / "datamarts"

files = [
    DM / "ml_linear_regression_results.csv",
    DM / "ml_ridge_results.csv",
    DM / "ml_random_forest_results.csv",
]

parts = []
for f in files:
    if f.exists():
        parts.append(pd.read_csv(f))
    else:
        print(f"⚠️ Fichier manquant (exécute d'abord le notebook correspondant) : {f.name}")

if not parts:
    raise FileNotFoundError("Aucun résultat trouvé. Lance les notebooks 01, 02 et 03.")

comparison = pd.concat(parts, ignore_index=True).sort_values("RMSE")
print("=== Tableau comparatif ===")
print(comparison.to_string(index=False))

best = comparison.iloc[0]
print("\n=== Modèle retenu ===")
print(f"Modèle : {best['model']}")
print(f"RMSE   : {best['RMSE']:.4f}")
print(f"MAE    : {best['MAE']:.4f}")

out = DM / "ml_model_comparison.csv"
comparison.to_csv(out, index=False)
print(f"\n✅ Export global : {out}")
```

## Justification pour la soutenance

1. **Alignement problématique** : les 3 modèles ciblent `taux_participation_reel`.
2. **Comparaison objective** : RMSE + MAE en Leave-One-Out.
3. **Choix final** : modèle au RMSE minimal, en rappelant les limites de qualité et de millésime des variables explicatives.
4. **Suite projet** : enrichir le datamart (grain département) pour renforcer la prédiction 2027.
